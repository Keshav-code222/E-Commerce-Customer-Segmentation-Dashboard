```python
df['Total'] = df['Quantity'] * df['UnitPrice']
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['InvoiceDate'].head()
```

```python
df.head()
```

```python
max_date = df['InvoiceDate'].max()
```

```python
max_date
```

```python
last_purchase = df.groupby('CustomerID')['InvoiceDate'].max()
```

```python
last_purchase
```

```python
recency = (max_date - last_purchase).dt.days
```

```python
recency
```

```python
frequency = df.groupby('CustomerID')['InvoiceNo'].nunique()
```

```python
monetary = df.groupby('CustomerID')['Total'].sum()
```

```python
rfm = pd.DataFrame({
    'Recency' : recency,
    'Frequency' : frequency,
    'Monetary' : monetary
})
```

```python
rfm.head()
```

```python
rfm = rfm[rfm['Monetary'] > 0]
```

```python
print(rfm.shape)
```

```python
rfm.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
rfm_scaled = scaler.fit_transform(rfm[['Recency', 'Frequency', 'Monetary']])
```

```python
print(rfm_scaled[:5])
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
inertia = []
K = range(1,11)
```

```python
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
clusters = kmeans.fit_predict(rfm_scaled)
```

```python
rfm['Cluster'] = clusters
```

```python
print(rfm.head())
print(rfm['Cluster'].value_counts())
```

```python
cluster_summary = rfm.groupby('Cluster')[['Recency', 'Frequency', 'Monetary']].mean()
print(cluster_summary)
```

```python
cluster_names = {
    0: 'Lost Customers',
    1: 'Regular Customers',
    2: 'Champions',
    3: 'Big Spenders'
}
```

```python
rfm['Cluster_Name'] = rfm['Cluster'].map(cluster_names)
```

```python
print(rfm['Cluster_Name'].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Cluster_Name', data=rfm, palette='viridis')
plt.title('Number of Customers in Each Cluster')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Cluster_Name', y='Monetary', data=rfm, palette='viridis', estimator='mean')
plt.title('Average Monetary Value by Cluster')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Recency', y='Monetary', hue='Cluster_Name', data=rfm, palette='viridis')
plt.title('Recency vs Monetary by Cluster')
plt.show()
```
